```sql
CREATE OR REFRESH STREAMING TABLE employee_raw_records 
(CONSTRAINT valid_date EXPECT (joined_date > to_date('01-JAN-2015', 'dd-MMM-yyyy')) ON VIOLATION DROP ROW)
AS SELECT *
FROM STREAM(employees);

CREATE OR REFRESH STREAMING TABLE flagged_employees 
AS SELECT
  date_trunc("week", joined_date) as joined_week,
  employee_id as employee_id, 
  employee_name
FROM
  STREAM(LIVE.employee_raw_records);

CREATE OR REFRESH MATERIALIZED VIEW employee_stats
AS SELECT
  date_trunc("week", joined_week) as week,
  count(*) as number_of_employees
FROM
 live.flagged_employees
GROUP BY date_trunc("week", joined_week);


```
